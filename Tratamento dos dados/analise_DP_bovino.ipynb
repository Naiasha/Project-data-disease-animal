```python
# fica somente com a células que não tem zero no número de animais acometidos
df_dp_1 = df_dp[df_dp['NUMERO_ANIMAIS_ACOMETIDOS'] != 0]
print(df_dp_1)
```

```python
df_dp_1_especies = df_dp_1['ESPECIE'].unique()
print(df_dp_1_especies)
```

```python
#filtrar pelas espécies referente a pesquisa - bovinos e suínos - e padronizar os nomes
df_dp_2 = df_dp_1[
    df_dp_1["ESPECIE"].isin(["Bovinos", "Suinos", "Bovino", "Suíno", "Bubalino"])
].copy()
df_dp_2["ESPECIE"] = df_dp_2["ESPECIE"].replace({
    "Bovinos": "Bovino",
    "Suinos": "Suíno",
    "Bubalinos": "Bubalino"
})
print(df_dp_2)
```

```python
df_bovino_anos = df_dp_2.groupby('MES_ANO', as_index=False)["NUMERO_ANIMAIS_ACOMETIDOS"].sum()
df_bovino_anos
```

```python
#separa os dados por espécie
df_bovino_bubalino = df_dp_2[df_dp_2['ESPECIE'].isin(['Bovino', 'Bubalino'])]
print(df_bovino_bubalino)
```

```python
df_bovino_bubalino_ano = df_bovino_bubalino.groupby('MES_ANO', as_index=False)["NUMERO_ANIMAIS_ACOMETIDOS"].sum()
df_bovino_bubalino_ano
```
