```python
#identificar colunas com not a number (NaN)
df_condenacao.isna().sum()
#identificar linhas que contenham algum not a number (NaN)
df_condenacao[df_condenacao.isna().any(axis=1)]
```

```python
df_condenacao['TOTAL_PARTES_AFETADAS'] = df_condenacao['TOTAL_PARTES_AFETADAS'].astype(int)
print(df_condenacao.info())
```

```python
df_condenacao_categorias = df_condenacao["ESPECIE"].unique()
sorted(df_condenacao_categorias)
```

```python
df_condenacao_categorias = df_condenacao["PARTE_AFETADA"].unique()
sorted(df_condenacao_categorias)
```

```python
#verifica a quantidade de informados 0
df_condenacao_0 = (df_condenacao == 0).sum()
print(df_condenacao_0)
```

```python
# fica somente com a células que não tem zero no número de animais acometidos
df_condenacao_1 = df_condenacao[df_condenacao['TOTAL_PARTES_AFETADAS'] != 0]
print(df_condenacao_1)
```

```python
#analisa para encontrar linhas duplicadas
duplicate_condenacao = df_condenacao[df_condenacao.duplicated()]
print(duplicate_condenacao) #resultado sem linhas duplicadas
```

```python
df_condenacao_suino = df_condenacao[df_condenacao['ESPECIE'].isin(['Suinos','Suíno'])].copy()
print(df_condenacao_suino)
```

```python
df_condenacao_suino_ano = df_condenacao_suino.loc[df_condenacao_suino["MES_ANO"].notnull()].copy()
df_condenacao_suino_ano["MES_ANO"] = pd.to_datetime(df_condenacao_suino_ano["MES_ANO"], format="%m/%Y", errors='coerce')
df_condenacao_suino_ano["ANO"] = df_condenacao_suino_ano["MES_ANO"].dt.year
df_condenacao_suino_ano["MES"] = df_condenacao_suino_ano["MES_ANO"].dt.month
print(df_condenacao_suino_ano.head(3))
```

```python
caminho = "C:/Users/thami/OneDrive/Desktop/Projeto PPDI/Dados/tratados"

df_condenacao_suino_ano.to_parquet(caminho + "/suino_ano.parquet", index=False)
print("Arquivos salvos com sucesso!")
```
