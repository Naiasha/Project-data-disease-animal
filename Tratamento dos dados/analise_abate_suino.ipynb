```python
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as pl #importa o pacote de plotagem
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 400
import seaborn as sns
import numpy as np
import plotly.express as px
```

```python
df_abate = pd.read_csv('C:/Users/thami/OneDrive/Documentos/Unitoledo/Pesquisa e produtividade/dados/qtt de animais abatidos por categoria.csv', 
                 sep=None,          # Let pandas detect the separator
                 engine='python',   # Use the python engine which is more flexible
                 encoding='utf-8',  # Specify encoding to handle special characters
                 on_bad_lines='warn')  # Use on_bad_lines instead of error_bad_lines
print(df_abate.head())
```

```python
df_abate.info()
```

```python
pd.set_option('display.float_format', '{:.0f}'.format) #0f é para sem casa decimal
print(df_abate.describe())
```

```python
#identificar colunas com not a number (NaN)
df_abate.isna().sum()
#identificar linhas que contenham algum not a number (NaN)
df_abate[df_abate.isna().any(axis=1)]
```

```python
df_abate = df_abate.dropna()
```

```python
df_abate['QUANTIDADE'] = df_abate['QUANTIDADE'].astype(int)
print(df_abate.info())
```

```python
#identificar valores únicos na coluna CATEGORIA e ordenar em ordem alfabética
df_abate_categorias = df_abate["CATEGORIA"].unique()
sorted(df_abate_categorias)
```

```python
#verifica a quantidade de informados 0
df_abate_0 = (df_abate == 0).sum()
print(df_abate_0)
```

```python
#analisa para encontrar linhas duplicadas
duplicate_abate = df_abate[df_abate.duplicated()]
print(duplicate_abate) #resultado sem linhas duplicadas
```

```python
# fica somente com a células que não tem zero no número de animais acometidos
df_abate_1 = df_abate[df_abate['QUANTIDADE'] != 0]
print(df_abate_1)
```

```python
df_abate_suino = df_abate[df_abate['CATEGORIA'].isin(['Suino Femea',
                                                                'Suino Macho',
                                                                'Suino Tipo Banha, Femea',
                                                                'Suino Tipo Banha, Macho',
                                                                'Suino Tipo Carne, Femea',
                                                                'Suino Tipo Carne, Macho',
                                                                'Suíno'
])].copy()
print(df_abate_suino)
```

```python
#Separar a coluna CATEGORIA em duas colunas: CATEGORIA e CATEG
#df_abate_bovinos_bubalinos [['CATEGORIA', 'CATEGORIA_SEX']] = df_abate_bovinos_bubalinos['CATEGORIA'].str.split(' ', n=1, expand=True)
#print(df_abate_bovinos_bubalinos)
```

```python
df_abate_suino.insert(loc=5, column= 'Sexo', value =['Femea' if 'Femea' in categoria 
                                                     else 'Macho' if 'Macho' in categoria
                                                     else 'Indeterminado' 
                                                     for categoria in df_abate_suino['CATEGORIA']])
print(df_abate_suino)
```

```python
df_abate_suino.insert(loc=5, column= 'ESPECIE', value =['Suíno' if 'Suino Femea' in categoria 
                                                     else 'Suíno' if 'Suino Macho' in categoria
                                                     else 'Suíno' if 'Suino Tipo Banha, Femea' in categoria
                                                     else 'Suíno' if 'Suino Tipo Banha, Macho' in categoria
                                                     else 'Suíno' if 'Suino Tipo Carne, Femea' in categoria
                                                     else 'Suíno' if 'Suino Tipo Carne, Macho' in categoria
                                                     else 'Suíno' 
                                                     for categoria in df_abate_suino['CATEGORIA']])
print(df_abate_suino)
```

```python
df_abate_suino.groupby(['CATEGORIA', 'ESPECIE'], as_index=False)['QUANTIDADE'].sum()
```

```python
df_abate_suino = df_abate[df_abate['CATEGORIA'].isin(['Leitao','Leitoa', 'Porca Matrizeira', 'Porco Matrizeiro', 'Porco Nao Classificado',
                                                      'Suino Femea', 'Suino Macho', 'Suino Tipo Banha  Femea', 'Suino Tipo Banha  Macho', 'Suino Tipo Carne  Femea',
                                                      'Suino Tipo Carne  Macho', 'Suíno'
])].copy()
print(df_abate_suino)
```

```python
df_abate_suino.insert(loc=5, column= 'Sexo', value =['Femea' if 'Femea' in categoria 
                                                     else 'Macho' if 'Macho' in categoria
                                                     else 'Femea' if 'Porca' in categoria
                                                     else 'Macho' if 'Porco' in categoria
                                                     else 'Femea' if 'Leitoa' in categoria
                                                     else 'Macho' if 'Leitao' in categoria
                                                     else 'Indeterminado' 
                                                     for categoria in df_abate_suino['CATEGORIA']])
print(df_abate_suino)
```

```python
df_abate_suino.insert(loc=6, column= 'ESPECIE', value =['Suíno' for _ in range(len(df_abate_suino))])
print(df_abate_suino)
```

```python
df_abate_suino_SE = df_abate_suino[df_abate_suino['UF_PROCEDENCIA'].isin(['SP', 'RJ', 'MG', 'ES'
])]
print(df_abate_suino_SE)
```
