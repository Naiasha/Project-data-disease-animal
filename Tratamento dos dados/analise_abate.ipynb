```python
df_abate = pd.read_csv('C:/Users/thami/OneDrive/\Documentos/Unitoledo/Pesquisa e produtividade/dados/qtt de animais abatidos por categoria.csv', 
                 sep=None,          # Let pandas detect the separator
                 engine='python',   # Use the python engine which is more flexible
                 encoding='utf-8',  # Specify encoding to handle special characters
                 on_bad_lines='warn')  # Use on_bad_lines instead of error_bad_lines
print(df_abate.head())
```

```python
df_abate['QUANTIDADE'] = df_abate['QUANTIDADE'].astype(int)
print(df_abate.info())
```

```python
#identificar colunas com not a number (NaN)
df_abate.isna().sum()
#identificar linhas que contenham algum not a number (NaN)
df_abate[df_abate.isna().any(axis=1)]
```

```python
df_abate = df_abate.dropna()
```

```python
#Analisando o ranking por estado de procedência
ranking_uf = (
    df_abate.groupby("UF_PROCEDENCIA", as_index=False)["QUANTIDADE"]
      .sum()
      .sort_values("QUANTIDADE", ascending=False)
)

ranking_uf
```
