```python
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as pl #importa o pacote de plotagem
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 400
```

#Leitura do arquivo Quantidade de doenças por procedência (QDP)

```python
df_2 = pd.read_csv('C:/Users/thami/OneDrive/Documentos/Unitoledo/Pesquisa e produtividade/dados/qnt de doenças por procedência.csv', 
                 sep=None,          # Let pandas detect the separator
                 engine='python',   # Use the python engine which is more flexible
                 encoding='utf-8',  # Specify encoding to handle special characters
                 on_bad_lines='warn')  # Use on_bad_lines instead of error_bad_lines
print(df_2.head())
```

```python
df_2.shape
```

```python
df_2.nunique
```

```python
df_2['QUANTIDADE'].sort_values().head(20)
```

```python
df_2['QUANTIDADE'].sort_values().tail(20)
```

```python
df_2[['QUANTIDADE']].info()
```

```python
df_2['QUANTIDADE'].value_counts()
```

```python
maior_qtd_2 = df_2.groupby(['DIAGNOSTICO'])['QUANTIDADE'].max()
print(maior_qtd_2)

asc = df_2.groupby(['DIAGNOSTICO'])['QUANTIDADE'].asc()
print(asc)
```

```python
df_2_ordenado = df_2.sort_values(by='QUANTIDADE')
print(df_2_ordenado)
```

```python
df_2['QUANTIDADE'].describe()
```

```python
pl.figure(figsize=(10, 5))

pl.hist(df_2['DIAGNOSTICO'], bins=50, edgecolor='black')

pl.xlabel('Quantidade')
pl.ylabel('Frequência')
pl.title('Histograma da Quantidade em escala log')
pl.show()
```

```python
df_2_not0 = df_2[df_2['QUANTIDADE'] != 0]
df_2_ordenado_not0 = df_2_not0.sort_values(by='QUANTIDADE')
print(df_2_ordenado_not0)
```
