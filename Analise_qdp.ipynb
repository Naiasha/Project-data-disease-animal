```python
df_QDP_2.columns = df_QDP_2.columns.str.strip().str.upper()
print(df_QDP_2)
```

```python
#verificar número 0
df_QDP_3 = (df_QDP_2 == 0).sum()
print(df_QDP_3)
```

```python
df_QDP_3 = df_QDP_2[df_QDP_2['QUANTIDADE'] != 0]
print(df_QDP_3)
```

```python
#configura para aparecer o númeto inteiro, sem notação científica
pd.set_option('display.float_format', '{:.0f}'.format) #0f é para sem casa decimal
print(df_QDP_3.describe())
```

```python
#Padronizar todos para minusculo
df_QDP_3.loc[:,'DIAGNOSTICO'] = df_QDP_3['DIAGNOSTICO'].str.strip().str.lower()
print(df_QDP_3)
```

```python
#CONFERINDO O VALOR MÁXIMO DO DESCRIBE
df_QDP_M = df_QDP_3[['DIAGNOSTICO','QUANTIDADE']].max()
print(df_QDP_M)
```

```python
#corrigindo palavras iguais mais com espaço a mais
df_QDP_3.loc[:,'DIAGNOSTICO'] = df_QDP_3['DIAGNOSTICO'].replace('estado anormal / patológico não previsto', 'estado anormal/patológico não previsto')
```
