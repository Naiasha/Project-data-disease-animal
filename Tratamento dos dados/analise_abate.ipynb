```python
df_abate = pd.read_csv('C:/Users/thami/OneDrive/Documentos/Unitoledo/Pesquisa e produtividade/dados/qtt de animais abatidos por categoria.csv', 
                 sep=None,          # Let pandas detect the separator
                 engine='python',   # Use the python engine which is more flexible
                 encoding='utf-8',  # Specify encoding to handle special characters
                 on_bad_lines='warn')  # Use on_bad_lines instead of error_bad_lines
print(df_abate.head())
```

```python
#identificar colunas com not a number (NaN)
df_abate.isna().sum()
#identificar linhas que contenham algum not a number (NaN)
df_abate[df_abate.isna().any(axis=1)]
```

```python
df_abate = df_abate.dropna()
```

```python
df_abate['QUANTIDADE'] = df_abate['QUANTIDADE'].astype(int)
print(df_abate.info())
```

```python
#identificar valores únicos na coluna CATEGORIA e ordenar em ordem alfabética
df_abate_categorias = df_abate["CATEGORIA"].unique()
sorted(df_abate_categorias)
```

```python
#verifica a quantidade de informados 0
df_abate_0 = (df_abate == 0).sum()
print(df_abate_0)
```

```python
#analisa para encontrar linhas duplicadas
duplicate_abate = df_abate[df_abate.duplicated()]
print(duplicate_abate) #resultado sem linhas duplicadas
```

```python
# fica somente com a células que não tem zero no número de animais acometidos
df_abate_1 = df_abate[df_abate['QUANTIDADE'] != 0]
print(df_abate_1)
```

```python
df_abate_bovinos_bubalinos = df_abate[df_abate['CATEGORIA'].isin(['Bovino','Bovino Femea','Bovino Macho','Bubalino','Bubalino Jovem','Bufala','Bufalo','Novilha Intermediaria',
                                                      'Novilha Precoce','Novilhao','Novilho Intermediario','Novilho Precoce','Novilhona', 'Vaca','Vitelo'
])].copy()
print(df_abate_bovinos_bubalinos)
```

```python
#Separar a coluna CATEGORIA em duas colunas: CATEGORIA e CATEG
#df_abate_bovinos_bubalinos [['CATEGORIA', 'CATEGORIA_SEX']] = df_abate_bovinos_bubalinos['CATEGORIA'].str.split(' ', n=1, expand=True)
#print(df_abate_bovinos_bubalinos)
```

```python
df_abate_bovinos_bubalinos.insert(loc=5, column= 'Sexo', value =['Femea' if 'Femea' in categoria 
                                                     else 'Macho' if 'Macho' in categoria
                                                     else 'Femea' if 'Vaca' in categoria
                                                     else 'Femea' if 'Novilha' in categoria
                                                     else 'Macho' if 'Novilho' in categoria
                                                     else 'Macho' if 'Novilhao' in categoria
                                                     else 'Femea' if 'Novilhona' in categoria
                                                     else 'Femea' if 'Bufala' in categoria
                                                     else 'Macho' if 'Bufalo' in categoria
                                                     else 'Indeterminado' 
                                                     for categoria in df_abate_bovinos_bubalinos['CATEGORIA']])
print(df_abate_bovinos_bubalinos)
```

```python
#df_abate_bovinos_bubalinos.insert(loc=6, column= 'Espécie', value =['Bubalino' if 'Bubalino' in categoria else 'Bovino' for categoria in df_abate_bovinos_bubalinos['CATEGORIA']])
#print(df_abate_bovinos_bubalinos)

df_abate_bovinos_bubalinos.insert(loc=5, column= 'ESPECIE', value =['Bubalino' if 'Bubalino' in categoria 
                                                     else 'Bubalino' if 'Bubalino Jovem' in categoria
                                                     else 'Bubalino' if 'Bufala' in categoria
                                                     else 'Bubalino' if 'Bufalo' in categoria
                                                     else 'Bovino' 
                                                     for categoria in df_abate_bovinos_bubalinos['CATEGORIA']])
print(df_abate_bovinos_bubalinos)
```

```python
df_abate_bovinos_bubalinos.groupby(['ANO', 'MES', 'CATEGORIA', 'ESPECIE'], as_index=False)['QUANTIDADE'].sum()
```

```python
df_abate_bovinos_bubalinos_SE = df_abate_bovinos_bubalinos[df_abate_bovinos_bubalinos['UF_PROCEDENCIA'].isin(['SP', 'RJ', 'MG', 'ES'
])]
print(df_abate_bovinos_bubalinos_SE)
```
