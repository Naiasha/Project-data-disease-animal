```python
df_condenacao['TOTAL_PARTES_AFETADAS'] = df_condenacao['TOTAL_PARTES_AFETADAS'].astype(int)
print(df_condenacao.info())
```

```python
df_condenacao_categorias = df_condenacao["ESPECIE"].unique()
sorted(df_condenacao_categorias)
```

```python
#verifica a quantidade de informados 0
df_condenacao_0 = (df_condenacao == 0).sum()
print(df_condenacao_0)
```

```python
# fica somente com a células que não tem zero no número de animais acometidos
df_condenacao_1 = df_condenacao[df_condenacao['TOTAL_PARTES_AFETADAS'] != 0]
print(df_condenacao_1)
```

```python
#analisa para encontrar linhas duplicadas
duplicate_condenacao = df_condenacao[df_condenacao.duplicated()]
print(duplicate_condenacao) #resultado sem linhas duplicadas
```

```python
df_condenacao_bovinos_bubalinos = df_condenacao[df_condenacao['ESPECIE'].isin(['Bovino','Bovino Femea','Bovino Macho','Bubalino','Bubalino Jovem','Bufala','Bufalo','Novilha Intermediaria',
                                                      'Novilha Precoce','Novilhao','Novilho Intermediario','Novilho Precoce','Novilhona', 'Vaca','Vitelo'
])].copy()
print(df_condenacao_bovinos_bubalinos)
```

```python
#Agrupa pela coluna ESPECIE e soma os valores da coluna TOTAL_PARTES_AFETADAS
df_condenacao_bovinos_bubalinos.groupby([ 'MES_ANO', 'ESPECIE'], as_index=False)['TOTAL_PARTES_AFETADAS'].sum()
```

```python
df_condenacao_bovinos_bubalinos_SE = df_condenacao_bovinos_bubalinos[df_condenacao_bovinos_bubalinos['UF_PROCEDENCIA'].isin(['SP', 'RJ', 'MG', 'ES'
])]
print(df_condenacao_bovinos_bubalinos_SE)
```
